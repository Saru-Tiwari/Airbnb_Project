```python
# First, install the Transformers library (if not already installed)
!pip install transformers

from transformers import pipeline

```

#  Load the summarization pipeline

```python

summarizer = pipeline("summarization")
```

#
# Sample text to summarize

```python
sample_text = """
I have always been passionate about technology and problem-solving, which led me to pursue a career in computer engineering.
Throughout my academic journey, I have gained strong technical skills in programming, software development, and system design.
 Participating in various projects, especially in app development and machine learning,
  has helped me understand real-world challenges and how to create effective solutions.
  These experiences have shaped my ability to work both independently and in teams, preparing me for the demands of the tech industry.

Looking ahead, I envision myself contributing to innovative projects that make a positive impact on society.
 I aim to work in a dynamic environment where I can continue learning and improving, possibly in fields like artificial intelligence,
  software development, or smart transportation systems. My long-term goal is to grow into a leadership role, where I can guide teams, share knowledge,
  and help shape the future of technology. I also plan to continue my education and stay updated with emerging trends to ensure
   I remain competitive and capable in this ever-evolving field.
"""
```

#  Function to summarize the text

```python
def summarize_text(text, max_length=100, min_length=30, do_sample=False):
    summary = summarizer(text, max_length=max_length, min_length=min_length, do_sample=do_sample)
    return summary[0]['summary_text']
```

# Generate summary

```python
summarized_text = summarize_text(sample_text)
print("Summarized Text:")
print(summarized_text)
```
